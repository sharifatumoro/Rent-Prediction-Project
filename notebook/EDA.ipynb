# 3. Statistical Analysis

```python
df.describe()
```

**Insights**

**.** The average rental price is approximately GH₵7,999, while the median is GH₵3,500. This suggests a positively skewed distribution, potentially influenced by high-priced properties.

**.** Properties typically have 2 bedrooms and 2 bathrooms, with average values of approximately 2.46 and 2.56, respectively.

**.** The median floor area is 145, considerably lower than the mean of 556.09, suggesting the presence of unusually large properties that may be potential outliers.

**.** Rental prices and floor area exhibit considerable variability, as indicated by their high standard deviations.

**.** The maximum values of 20 bedrooms, 20 bathrooms, and 5,000 floor area suggest potential extreme observations.

# 4. Univariate Analysis

```python
plt.figure(figsize=(10, 6))

sns.histplot(df['price'], bins=50, kde=True)

plt.title('Distribution of House Rental Prices')
plt.xlabel('Rental Price')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df['price'])

plt.title('Boxplot of Rental Prices')
plt.xlabel('Rental Price')

plt.show()
```

### **a.** for numerical analysis

```python
plt.figure(figsize=(10,8))
sns.histplot(df, x='price', bins=50, kde=True)
plt.title("Distribution of Price")
plt.show()
```

It is observed that rent in Ghana ranges form ₵1,000.00 to ₵60,000.00 with majority below ₵10,000.00 per month. The distribution is positively skewed meaning most of the properties are concentrated within lower rental price range.
